# 3. Select two features (e.g., alcohol and density) from the dataset and calculate the covariance matrix using np.cov(X.T), where X is the feature matrix consisting of the selected columns.(Score 2)

```python
X = df[['alcohol','density']].values
```

```python
covariance = np.cov(X.T)
```

```python
covariance
```

# 4. Perform eigen decomposition on the covariance matrix you computed in question 3. Identify and interpret the results:Identify the top 2 eigenvalues of the covariance matrix,Identify the corresponding eigenvectors.(Score 2)

```python
eigen_val, eigen_vector = np.linalg.eig(covariance)
```

```python
for e in eigen_val:
    print(e)
```
