```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

titanic = sns.load_dataset("titanic")
```

```python
titanic
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=titanic, x='embarked', hue='sex')
plt.title("Passenger Count by Embarkation Port (Sex-wise)")
plt.xlabel("Embarked Port")
plt.ylabel("Passenger Count")
plt.legend(title="Sex")
plt.show()
```

Passengers mostly boarded from port S, followed by C, and least from Q.
Across all ports, males are more frequent than females.

```python
plt.figure(figsize=(10,6))
sns.swarmplot(data=titanic, x='pclass', y='age', hue='sex')
plt.title("Age Distribution by Passenger Class and Sex")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.legend(title="Sex")
plt.show()
```

Higher classes (1st) tend to have older passengers, while lower classes (3rd) have a wider spread of younger ages. Females show slightly more concentration in younger age ranges.

```python
plt.figure(figsize=(6,6))
gender_counts = titanic['sex'].value_counts()

plt.pie(gender_counts,
        labels=gender_counts.index,
        autopct='%1.1f%%',
        explode=[0.05, 0], 
        startangle=90)

plt.title("Gender Distribution of Passengers")
plt.show()
```

The Titanic dataset contains more male passengers than female passengers, with males making up about 65% of the dataset.

```python
corr_data = titanic[['age','fare','pclass','sibsp','parch','survived']].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr_data, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

1. Fare is negatively correlated with pclass (higher class = higher fare).
2. Survival has a slight positive relation with fare and negative relation with pclass.
3. Family size variables (sibsp, parch) have moderate correlation with each other

```python
plt.figure(figsize=(10,6))
sns.violinplot(data=titanic, x='pclass', y='fare', hue='sex', split=True)
plt.title("Fare Distribution by Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("age")
plt.legend(title="Sex")
plt.show()
```

Higher classes (especially 1st class) have a much broader spread of fares, showing both very high and moderate ticket prices. Females in first class appear to have slightly higher median fares.

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=titanic, x='age', y='fare', hue='survived', style='survived')
plt.title("Age vs Fare (Colored by Survival Status)")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived")
plt.show()
```
