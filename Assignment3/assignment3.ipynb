```python
arr2 = np.arange(1, 21)
extracted = arr2[5:16]  
print(extracted)
```

```python
mean_val = np.mean(arr2)
median_val = np.median(arr2)
std_val = np.std(arr2)

print("Mean:", mean_val)
print("Median:", median_val)
print("Standard Deviation:", std_val)
```
