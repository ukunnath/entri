```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
import pickle

# -----------------------------
# 1️⃣ Split features and target
# -----------------------------
target_col = 'default_payment_next_month'
X = df.drop(target_col, axis=1)
y = df[target_col]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# -----------------------------
# 2️⃣ Identify numeric & categorical columns
# -----------------------------
numeric_features = X_train.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_features = X_train.select_dtypes(include=['object', 'category']).columns.tolist()

# Convert categorical columns to object type to allow numeric mapping
for col in categorical_features:
    X_train[col] = X_train[col].astype('object')
    X_test[col] = X_test[col].astype('object')

# -----------------------------
# 3️⃣ Target (mean) encoding for categorical columns
# -----------------------------
target_encoders = {}
for col in categorical_features:
    mapping = X_train.join(y_train).groupby(col)[target_col].mean().to_dict()
    target_encoders[col] = mapping
    # Apply mapping to train
    X_train[col] = X_train[col].map(mapping)

# Apply mapping to test (unseen categories replaced by overall mean)
overall_mean = y_train.mean()
for col, mapping in target_encoders.items():
    X_test[col] = X_test[col].map(mapping).fillna(overall_mean)

# Save target encoders for deployment
with open('target_encoders.pkl', 'wb') as f:
    pickle.dump(target_encoders, f)

# -----------------------------
# 4️⃣ Preprocessing pipeline for numeric columns
# -----------------------------
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        # categorical columns are already encoded numerically
    ],
    remainder='passthrough'  # leave target-encoded categorical columns as-is
)

# -----------------------------
# 5️⃣ Define models to compare
# -----------------------------
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=200, random_state=42),
    'Gradient Boosting': GradientBoostingClassifier(n_estimators=200, random_state=42),
    'XGBoost': XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42),
    'LightGBM': LGBMClassifier(random_state=42)
}

# -----------------------------
# 6️⃣ Evaluate models with Stratified K-Fold CV
# -----------------------------
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
results = {}

for name, model in models.items():
    pipe = Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])
    
    scores = cross_val_score(pipe, X_train, y_train, cv=cv, scoring='roc_auc')
    results[name] = scores.mean()

# -----------------------------
# 7️⃣ Display ROC_AUC table sorted descending
# -----------------------------
results_df = pd.DataFrame(list(results.items()), columns=['Model', 'Mean_ROC_AUC'])
results_df = results_df.sort_values(by='Mean_ROC_AUC', ascending=False).reset_index(drop=True)
print("Cross-validated ROC_AUC scores:")
print(results_df)

# -----------------------------
# 8️⃣ Fit best model on full training set
# -----------------------------
best_model_name = results_df.loc[0, 'Model']
best_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', models[best_model_name])
])
best_model.fit(X_train, y_train)

# Save best model for deployment
with open('best_model.pkl', 'wb') as f:
    pickle.dump(best_model, f)

# -----------------------------
# 9️⃣ Evaluate on test set
# -----------------------------
y_pred_proba = best_model.predict_proba(X_test)[:,1]
from sklearn.metrics import roc_auc_score
test_roc_auc = roc_auc_score(y_test, y_pred_proba)
print(f"Test ROC_AUC for {best_model_name}: {test_roc_auc:.4f}")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.base import BaseEstimator, TransformerMixin
import pickle

# -----------------------------
# 1️⃣ Define CV-safe Target Encoder with smoothing
# -----------------------------
class TargetEncoder(BaseEstimator, TransformerMixin):
    def __init__(self, smoothing=10.0):
        self.smoothing = smoothing
        self.mapping = {}
        self.overall_mean = None
    
    def fit(self, X, y):
        self.overall_mean = y.mean()
        for col in X.columns:
            stats = X.join(y).groupby(col)[y.name].agg(['mean','count'])
            smooth = (stats['mean'] * stats['count'] + self.overall_mean * self.smoothing) / (stats['count'] + self.smoothing)
            self.mapping[col] = smooth.to_dict()
        return self
    
    def transform(self, X):
        X_encoded = X.copy()
        for col in X_encoded.columns:
            X_encoded[col] = X_encoded[col].map(self.mapping[col]).fillna(self.overall_mean)
        return X_encoded

# -----------------------------
# 2️⃣ Split dataset into train/test
# -----------------------------
target_col = 'default_payment_next_month'
X = df.drop(target_col, axis=1)
y = df[target_col]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# -----------------------------
# 3️⃣ Identify numeric & categorical columns
# -----------------------------
numeric_features = X_train.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_features = X_train.select_dtypes(include=['object', 'category']).columns.tolist()

# Convert categorical columns to object
for col in categorical_features:
    X_train[col] = X_train[col].astype('object')
    X_test[col] = X_test[col].astype('object')

# -----------------------------
# 4️⃣ Define preprocessing pipeline
# -----------------------------
target_encoder = TargetEncoder(smoothing=10)

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', target_encoder, categorical_features)
    ]
)

# -----------------------------
# 5️⃣ Define models
# -----------------------------
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=200, random_state=42),
    'Gradient Boosting': GradientBoostingClassifier(n_estimators=200, random_state=42),
    #'XGBoost': XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42),
    #'XGBoost': XGBClassifier(max_depth=2,gamma=10,colsample_bytree=0.4),
    'XGBoost': XGBClassifier(n_estimators=500,max_depth=4,learning_rate=0.05,subsample=0.8,colsample_bytree=0.8,reg_lambda=0.5,eval_metric='auc', random_state=42),
    'LightGBM': LGBMClassifier(random_state=42)
}

# -----------------------------
# 6️⃣ Evaluate models with Stratified K-Fold CV
# -----------------------------
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
results = {}

for name, model in models.items():
    pipe = Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])
    scores = cross_val_score(pipe, X_train, y_train, cv=cv, scoring='roc_auc')
    results[name] = scores.mean()

# Create sorted ROC_AUC table
results_df = pd.DataFrame(list(results.items()), columns=['Model', 'Mean_ROC_AUC'])
results_df = results_df.sort_values(by='Mean_ROC_AUC', ascending=False).reset_index(drop=True)
print("Cross-validated ROC_AUC scores:")
print(results_df)

# -----------------------------
# 7️⃣ Fit best model on full training set
# -----------------------------
best_model_name = results_df.loc[0, 'Model']
best_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', models[best_model_name])
])
best_model.fit(X_train, y_train)

# -----------------------------
# 8️⃣ Evaluate on test set
# -----------------------------
y_pred_proba = best_model.predict_proba(X_test)[:,1]
from sklearn.metrics import roc_auc_score
test_roc_auc = roc_auc_score(y_test, y_pred_proba)
print(f"Test ROC_AUC for {best_model_name}: {test_roc_auc:.4f}")
print(f"Training ROC_AUC for {best_model_name}: { roc_auc_score(y_train, best_model.predict_proba(X_train)[:,1]):.4f}")

# -----------------------------
# 9️⃣ Save best model & target encoders for deployment
# -----------------------------
with open('best_model.pkl', 'wb') as f:
    pickle.dump(best_model, f)

with open('target_encoders.pkl', 'wb') as f:
    pickle.dump(target_encoder.mapping, f)
```

```python
# =========================================================
# 0. IMPORTS
# =========================================================

import pandas as pd
import numpy as np
import pickle

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.base import BaseEstimator, TransformerMixin

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import roc_auc_score

from catboost import CatBoostClassifier


# =========================================================
# 1. TARGET ENCODER (SMOOTHED, DEPLOYMENT SAFE)
# =========================================================

class TargetEncoder(BaseEstimator, TransformerMixin):
    def __init__(self, smoothing=10):
        self.smoothing = smoothing
        self.mapping = {}
        self.global_mean = None

    def fit(self, X, y):
        self.global_mean = y.mean()
        for col in X.columns:
            stats = y.groupby(X[col]).agg(['mean', 'count'])
            smooth = (
                stats['mean'] * stats['count']
                + self.global_mean * self.smoothing
            ) / (stats['count'] + self.smoothing)
            self.mapping[col] = smooth.to_dict()
        return self

    def transform(self, X):
        X = X.copy()
        for col in X.columns:
            X[col] = X[col].map(self.mapping[col]).fillna(self.global_mean)
        return X


# =========================================================
# 2. TRAIN / TEST SPLIT (ONCE)
# =========================================================

target = 'default_payment_next_month'

X = df.drop(columns=[target])
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)




num_cols = X_train.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = X_train.select_dtypes(include=['object', 'category']).columns.tolist()


for col in cat_cols:
    X_train[col] = X_train[col].astype('object')
    X_test[col] = X_test[col].astype('object')


# =========================================================
# 3. PREPROCESSORS
# =========================================================

# --- OneHot Encoding ---
ohe_preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), cat_cols)
    ]
)

# --- Target Encoding ---
te_preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols),
        ('cat', TargetEncoder(smoothing=10), cat_cols)
    ]
)


# =========================================================
# 4. MODELS
# =========================================================

models_ohe = {
    'LogReg_OHE': LogisticRegression(max_iter=1000),
    'RF_OHE': RandomForestClassifier(n_estimators=300, random_state=42)
}

models_te = {
    'LogReg_TE': LogisticRegression(max_iter=1000),
    'RF_TE': RandomForestClassifier(n_estimators=300, random_state=42),
    'GB_TE': GradientBoostingClassifier(n_estimators=300)
}


# =========================================================
# 5. MODEL COMPARISON (ROC_AUC)
# =========================================================

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
results = []

def evaluate_model(name, pipeline):
    score = cross_val_score(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring='roc_auc'
    ).mean()
    results.append({'Model': name, 'ROC_AUC': score})


# ---- OneHot models ----
for name, model in models_ohe.items():
    pipe = Pipeline([
        ('prep', ohe_preprocessor),
        ('model', model)
    ])
    evaluate_model(name, pipe)


# ---- Target Encoding models ----
for name, model in models_te.items():
    pipe = Pipeline([
        ('prep', te_preprocessor),
        ('model', model)
    ])
    evaluate_model(name, pipe)


# =========================================================
# 6. RESULTS TABLE (SORTED)
# =========================================================

results_df = (
    pd.DataFrame(results)
    .sort_values('ROC_AUC', ascending=False)
    .reset_index(drop=True)
)

print("\nMODEL COMPARISON (ROC_AUC):")
print(results_df)


# =========================================================
# 7. TRAIN BEST MODEL ON FULL TRAINING DATA
# =========================================================

best_model_name = results_df.loc[0, 'Model']
print("\nBest Model:", best_model_name)

if best_model_name == 'CatBoost':
    pass

else:
    if 'OHE' in best_model_name:
        preprocessor = ohe_preprocessor
        model = models_ohe[best_model_name]
    else:
        preprocessor = te_preprocessor
        model = models_te[best_model_name]

    final_model = Pipeline([
        ('prep', preprocessor),
        ('model', model)
    ])
    final_model.fit(X_train, y_train)


# =========================================================
# 8. FINAL TEST ROC_AUC
# =========================================================

if best_model_name == 'CatBoost':
    y_test_proba = final_model.predict_proba(X_test)[:, 1]
else:
    y_test_proba = final_model.predict_proba(X_test)[:, 1]

test_auc = roc_auc_score(y_test, y_test_proba)
print("\nTest ROC_AUC:", test_auc)


# =========================================================
# 9. SAVE MODEL FOR DEPLOYMENT
# =========================================================

with open('best_model.pkl', 'wb') as f:
    pickle.dump(final_model, f)

print("\nModel saved as best_model.pkl")
```
